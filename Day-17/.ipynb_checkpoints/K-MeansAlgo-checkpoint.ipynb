# UNSUPERVISED Machine Learning Technique

## K-Means Clustring Example

    - Terms we will use here like Cluster
    - Cluster: partitiondataset ino K distinct cluster
    - Centroid: nearest mean cluster


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans

import warnings #<--- to hide the pink warning box!
warnings.filterwarnings('ignore')
```

```python
X = np.array([
    [7.5, 8.0],
    [9.0, 7.0],
    [10.5, 6.5],
    [6.0, 9.0],
    [13.0, 5.5],
    [8.0, 8.5],
    [14.0, 4.5],
    [9.5, 7.0],
    [7.0, 9.0],
    [12.0, 6.0]
])
```

```python
df = pd.DataFrame(X, columns=['EntryTime','Duration'])
df
```

```python
# create model

model = KMeans(n_clusters=3, random_state=0) #<-- usually use odd number for cluster
```

```python
df['Cluster'] = model.fit_predict(df[['EntryTime','Duration']])
centers = model.cluster_centers_
df
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df['EntryTime'], df['Duration'], c=df['Cluster'])
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200)
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df['EntryTime' ], df['Duration'], c=df['Cluster' ], s=100)
plt.scatter(centers[:,0], centers[:,1], marker='X', s=200,c='red',label='Centroid')
plt.xlabel('Entry Time (Hour of Day)')
plt.ylabel('Duration in Office (Hours)')
plt.title('Working Hour Patterns using K-Means')
plt.legend()
plt.grid(True)
plt.show()
```

```python
X = train_df[['AnnualIncome','SpendingScore']]
kmeans = KMeans(n_clusters=5, random_state=0)
labels = kmeans.fit_predict(X)
train_df['Cluster'] = labels
centers = kmeans.cluster_centers_
train_df.head()
```

```python
plt.figure(figsize=(7,6))
plt.scatter(train_df['AnnualIncome'], train_df['SpendingScore'], c=train_df['Cluster'], alpha=0.85)
plt.xlabel('Annual Income (MYR)')
plt.ylabel('Spending Score (1-100)')
plt.title('Customer Segments (K-Means)')
plt.grid(True, linewidth=0.3)
plt.tight_layout()
plt.show()
```
