# Steps for checking Model
    - 1. Base Line (mean prediction)
    - 2. Data Cleaning & Preparation
    - 3. Linear Regression Model
    - 4. Feature Engineering
    - 5. Tuning
    - 6. Cross Validation
    - 7. Residual Analysis

```python
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = {
    'StudyHours': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Attendance': [60, 65, 70, 72, 75, 80, 85, 88, 90],
    'PrevScore': [40, 45, 50, 55, 60, 65, 70, 75, 80],
    'FinalMarks': [50, 55, 60, 65, 70, 75, 78, 85, 88]
}
```

```python
df=pd.DataFrame(data)
df
```

```python
y=df['FinalMarks']
print('FInal Marks as below: \n')
y
```

```python
y_pred_baseline=np.repeat(y.mean(),len(y))
```

```python
mae=mean_absolute_error(y,y_pred_baseline)
mae
```

```python
mse=mean_squared_error(y,y_pred_baseline)
mse
```

```python
rmse=np.sqrt(mse)
rmse
```

```python
r2 = r2_score(y,y_pred_baseline)
print(f'Baseline MAE={mae:.2f}, RMSE={rmse:.2f}, R-Squared={r2:.2f}')
```

```python
print(df.isnull().sum())
```

```python
df=df.fillna(df.mean())
df
```

```python
# Data Preparation, find input & output


X= df[['StudyHours','Attendance','PrevScore']]
y=df['FinalMarks']

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42) #<-- 0.3 = 30% of the model
```

```python
# Model Creation


model=LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
# Just to see outcome results of y orginal vs prediction


print('Original FInal Marks')
print(y)
y_all_prediction=model.predict(X)
print('Predicted all Final Marks')
print(y_all_prediction)
```

```python
y_pred=model.predict(X_test)
y_pred
```

```python
mae=mean_absolute_error(y_test,y_pred)
mae
```

```python
mse=mean_squared_error(y_test,y_pred)
mse
```

```python
rmse=np.sqrt(mse)
rmse
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
print(f'Baseline MAE={mae:.2f}, RMSE={rmse:.2f}, R-Squared={r2:.2f}')
```
