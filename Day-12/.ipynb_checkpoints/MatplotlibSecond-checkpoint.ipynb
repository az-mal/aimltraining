```python
import pandas as pd
df=pd.read_csv('patients.csv')
print(df.head())
```

```python
age_counts=df['age'].value_counts().sort_index()
plt.figure(figsize=(12,5))
plt.bar(age_counts.index,age_counts.values,color='skyblue',width=0.3,edgecolor='orange')
plt.title('Number of Patients by Age')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.show()
```

```python
print(df['service'].value_counts())
```

```python
avg_age_bydate=df.groupby('arrival_date')['age'].mean()
print(avg_age_bydate)
```

##### Ex: From above, create a bar plot - average age of patients by arrival date

```python
df['arrival_date']=pd.to_datetime(df['arrival_date'])
avg_age_bydate=df.groupby('arrival_date')['age'].mean()
plt.figure(figsize=(15,3))
plt.bar(avg_age_bydate.index.astype(str),avg_age_bydate.values,color='skyblue',width=0.3,edgecolor='orange')
plt.title('Average Age of Patients by Arrival Date')
plt.xlabel('Arrival Date')
plt.ylabel('Average Age')
plt.grid(axis='y',linestyle='--',alpha=0.3)
plt.show()
```

```python
print(df['service'])
```

```python
petients_by_service=df.groupby('service')['age'].mean()
print(petients_by_service)
```

#### Scatter Plots
    - In scatter plots we represent data in points , where each point represent a data observation , We can check how to values correlate

```python
product_sales={'Laptop':750,'Smartphone':650,'Smartwatch':400,'Tablet':500,'AC':400} #<--- use diff. bracket since input value
products=list(product_sales.keys())
sales=list(product_sales.values())
plt.scatter(products,sales,color='purple',marker='x',s=150)
plt.title("Basic Product and Sales Representation")
plt.xlabel("Products")
plt.ylabel('Sales')
plt.grid('True')
plt.show()
```

```python
players=['MSD','Rohot','Virat','Gill','Shami']
match1_score=[45,67,56,78,89]
match2_score=[55,78,40,80,97]
match3_score=[90,40,27,33,69]
plt.figure(figsize=(8,5))
plt.scatter(players,match1_score,color='red',label='First Match',s=100)
plt.scatter(players,match2_score,color='green',label='Second Match',s=100,marker='d')
plt.scatter(players,match3_score,color='blue',label='Third Match',s=100,marker='x')
plt.title('Comparison of Players in Two Matches')
plt.xlabel('\nPlayers')
plt.ylabel('Score\n')
plt.legend(loc='lower left')
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

##### Ex: Write a comparison that shows average calories burned for each workout type from final_data dataset

```python
df=pd.read_csv('Final_data.csv')
avg_cal=df.groupby('Workout_Type')['Calories_Burned'].mean()
selected_index=avg_cal.index
selected_value=avg_cal.values
plt.figure(figsize=(6,4))
plt.scatter(selected_index,selected_value,color='red',s=100)
plt.title('Comparison of calories burned by type of workouts')
plt.xlabel('\nWorkouts')
plt.ylabel('Calories Burned\n')
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

```python
python_lib=['Numpy','Pandas','Matplotlib','Scikit-learn']
liked_by=[5000,7000,9000,4000]
bar_colors=['red','blue','green','black']
plt.figure(figsize=(8,3))
plt.bar(python_lib,liked_by,color=bar_colors,width=0.25,edgecolor='yellow')
plt.title('Python Library Likes')
plt.xlabel('library')
plt.ylabel('likes')
plt.show()
```

```python
plt.figure(figsize=(7,2))
plt.barh(python_lib,liked_by,color=bar_colors,height=0.4,edgecolor='yellow')
plt.title('Python Library Likes')
plt.xlabel('library')
plt.ylabel('likes')
plt.show()
```
