```python
import numpy as np
import random
```

```python
empty_road=0
light_traffic=1
moderate_traffic=2
heavy_traffic=3
very_heavy_traffic=4

states = np.arange(0,5,1) # <-- lowest 0, highest 101, increment range 5
actions = ['GREEN','RED']
print('Level: ',states)
print('Actions: ',actions)
```

```python
print(states)
print(actions)
```

```python
Q=np.zeros((len(states),len(actions)))

alpha=0.1
gamma=0.9
epsilon=0.2
episodes=300

print('Q-table shape: ',Q.shape)
```

```python
def get_reward(level, action):
    if 3 <= level <= 4:
        reward = 10 # comfortable
    else:
        reward = -10 #too low/too high
    if action == 'GREEN' and level >=3:
        reward + 10 # overflow risk
    if action == 'RED' and level <= 2: 
        reward - 10 # empty risk
    return reward

print('Reward Example (level=4, GREEN):', get_reward(4, 'GREEN'))
```

```python
def next_level(level, action):
    if action=='':
        level+=random.choice([0,4,1])
    else:
        level-=random.choice([0,4,1])
    return int(np.clip(level,0,4))

print('Next Level Example: ',next_level(3,'GREEN'))
```

```python
for ep in range(episodes):
    level = random.choice(states)
    done = False
    for _ in range(1):   # Limit steps per episode
        # Choose action (epsilon-greedy)
        if random.uniform(0,1) < epsilon:
            action = random.choice(actions)
        else:
            action = actions[np.argmax(Q[level // 1])]

        next_state = next_level(level, action)
        reward = get_reward(next_state, action)
        a = actions.index(action)
        best_next = np.max(Q[next_state // 1])
        Q[level// 1, a] += alpha * (reward + gamma * best_next - Q[level // 1, a])

print('Training Completed')
```
