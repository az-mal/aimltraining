```python
df
```

```python
model.fit(X,y)
```

```python
new_cust_age=int(input('Enter your age: '))
new_cust_sal=int(input('Enter your salary: '))
profile=[[new_cust_age, new_cust_sal]]
prediction=model.predict(profile)
if (prediction):
    print(f'{profile[0][1]} Purchasing')
else:
    print('Not Purchasing')
```

## Exercise for Logistic Regression - Email Spam or Not dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
```

```python
df=pd.read_csv("spam_dataset.csv")
df.head(5)
```

```python
num_rows = len(df)
print(f"Number of rows: {num_rows}")
```

```python
print('Number of Missing Values are:',df.isnull().sum())
```

```python
X=df['message_content']
y=df['is_spam']

cv=CountVectorizer()
X_cv=cv.fit_transform(X)
```

```python
model=LogisticRegression()
```

```python
model.fit(X_cv,y)
```

```python
sample=str(input('Message Received: '))
prediction=model.predict(cv.transform(sample))
if(prediction.all()):
    print('Not Spam')
else:
    print('Spam')
```
