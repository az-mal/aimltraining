```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
```

```python
df_spam_email = pd.read_csv('spam_dataset.csv')
df_spam_email
```

```python
print('Number of Missing values are:',df_spam_email.isnull().sum())
```

```python
xSpam = df_spam_email['message_content']
ySpam = df_spam_email['is_spam']
```

```python
# Split the data (e.g., 80% for training, 20% for testing)
X_train_text, X_test_text, y_train, y_test = train_test_split(
    xSpam, 
    ySpam, 
    test_size=0.2, 
    random_state=42
)
print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))
```

```python
vectorizer = CountVectorizer()
X_train_numeric = vectorizer.fit_transform(X_train_text)
X_test_numeric = vectorizer.transform(X_test_text)
```

```python
model_spam_email = LogisticRegression() 
model_spam_email.fit(X_train_numeric, y_train)
```

```python
# new_email_text = ["Free gift! Click to win millions of dollars now."]
new_email_text = str(input('Enter Message: '))
new_email_numeric = vectorizer.transform([new_email_text]) 
prediction_spam = model_spam_email.predict(new_email_numeric)

if (prediction_spam):
    print('This email is likely spam')
else:
    print('This email is likely not spam')
```
