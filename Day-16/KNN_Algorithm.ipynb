```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

import warnings #<--- to hide the pink warning box!
warnings.filterwarnings('ignore')
```

```python
X = np.array([
    [1,3500], #unfit
    [2,3200], #unfit
    [3,3000], #unfit
    [4,2500], #fit
    [5,2300], #fit
    [6,2000], #fit
    [7,2200], #fit
    [8,3700], #unfit
])
y = ['Unfit','Unfit','Unfit','Fit','Fit','Fit','Fit','Unfit',]
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
new_person = [[4.5,2600]]
pred = model.predict(new_person)
print('Predicted Result: ',pred[0])
```

```python
exercise_hours=float(input("Enter number of exercise hours: "))
cal_intake=float(input("Enter number of calories consumed: "))
pred_fu=model.predict([[exercise_hours,cal_intake]])
print(f'Exercise Hours {exercise_hours}, Calories Intake {cal_intake} You are likely {pred_fu[0]}')                     
```

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv('fruit_data.csv')
df.head(5)
```

```python
num_rows = len(df)
print(f"Number of rows: {num_rows}")
```

```python
print('Number of Missing Values are:',df.isnull().sum())
```

```python
X = df[['Weight','Size']]
y= df['Fruit']
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
weight=float(input("Enter the fruit weight (g): "))
size=float(input("Enter the fruit size (cm): "))
prediction=model.predict([[weight,size]])
print(f'With weight of {weight} and size of fruit {size} very likely the fruit is {prediction[0]}')     
```

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv('student_grade.csv')
df.head(5)
```

```python
num_rows = len(df)
print(f"Number of rows: {num_rows}")
```

```python
print('Number of Missing Values are:',df.isnull().sum())
```

```python
X = df[['Hours_Studied','Assignments_Completed']]
y= df['Grade']
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(X,y)
```

```python
hours=float(input("Enter the hours allocated to study: "))
assignment=float(input("Enter how many assignment completed: "))
prediction=model.predict([[hours,assignment]])
print(f'With hours spent to study of {hours} and assignment completed {assignment} very likely to get grade {prediction[0]}')  
```
