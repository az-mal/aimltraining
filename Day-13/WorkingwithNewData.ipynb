```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('netflix_data.csv')
print(df.head())
```

```python
df=df.dropna(subset=['release_year'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['release_year'],bins=30,color='green',edgecolor='black',alpha=0.8)
plt.title('Netflix Released Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue',linestyle='--',linewidth=0.7)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)
plt.figure(figsize=(8,5))
plt.hist(df['release_year'],bins=100,color='green',edgecolor='yellow',alpha=0.8)
plt.title('Netflix Released Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue',linestyle='--',linewidth=0.7)
plt.tight_layout()
plt.savefig('netflix_bar.pdf',dpi=300,bbox_inches='tight')
plt.show()
```

#### Ex: Use netflix_data.csv dataset
    - print how many Movies and TV Shows are available on Netflix
    - Display Movies and TV Shows proportion in a pie plot
    - Save file as pdf

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('netflix_data.csv')
df=df.dropna(subset=['type'])
type_counts=df['type'].value_counts()
print(type_counts)
```

```python
plt.figure(figsize=(5,5))
plt.pie(type_counts,
        labels=type_counts.index,
        colors=['green','cyan'],
        autopct='%1.5f%%',
        startangle=140,
        shadow=True,
        textprops={'color':'black','fontsize':10},
        explode=(0.07,0),
       )
plt.legend(title='Movies/Shows')
plt.title('Netflix Movies and TV Shows Distribution')
plt.savefig('showsvsmovies_pie.pdf',dpi=300,bbox_inches='tight')
plt.tight_layout()
plt.show()
```
