```python
import pandas as pd
```

```python
data={'Students':['Amit','Anam','Chi','Sam','Ravi','Jai'], 
      'Marks':[85,None,65,75,None,55]}
```

```python
df=pd.DataFrame(data)
print('Original Data')
df
```

```python
# to find the missing data:

print("Missing Data")
print("Number of Missing Values Are:",df.isnull().sum())
```

```python
# to fill the missing data:

average_marks=df['Marks'].mean()
print(average_marks)
```

```python
df.fillna(average_marks, inplace=True)
```

```python
print('After fillna() call')
df
```

## dropna() example

```python
import pandas as pd  # Import pandas library first

students_data = {'Name': ['Amit', 'Anam', 'Chi', 'Sam', 'Ravi', 'Jai'], 'Marks': [85, None, 65, 75, None, 55]}
df = pd.DataFrame(students_data)  # Create DataFrame using pd.DataFrame, not df.pd.DataFrame
print("Original Data")
df  # Display the DataFrame
```

```python
df=df.dropna(subset=["Marks"])
```

```python
print("Dataset After dropna()")
df
```
